```python
import numpy as np

# Definindo a matriz de dados e os dias da semana
dados = np.array([
    [120, 15000, 100, 125.00, 4],
    [135, 16200, 108, 120.00, 5],
    [98, 11700, 82, 119.39, 3],
    [150, 21000, 125, 140.00, 7],
    [175, 24500, 140, 140.00, 9],
    [160, 22400, 132, 140.00, 6],
    [110, 13200, 95, 120.00, 2]
])

dias = np.array([
    "Segunda",
    "Terça",
    "Quarta",
    "Quinta",
    "Sexta",
    "Sábado",
    "Domingo"
])

indicadores = np.array(["Pedidos", "Faturamento", "Clientes", "Ticket_Medio", "Reclamações"])

# ==========================================
# Parte 1 — Conhecendo os dados
# ==========================================
print("--- Parte 1 — Conhecendo os dados ---")
print("Quantidade de linhas e colunas (.shape):", dados.shape)
print("Quantidade de linhas:", dados.shape[0])
print("Quantidade de colunas:", dados.shape[1])
print("Quantidade total de valores (.size):", dados.size)
print("Número de dimensões (.ndim):", dados.ndim)
print("Tipo dos dados armazenados (.dtype):", dados.dtype)

# ==========================================
# Parte 2 — Acessando os dados
# ==========================================
print("\n--- Parte 2 — Acessando os dados ---")
# 2.1 Dados referentes à sexta-feira (índice 4)
print("2.1 Dados da Sexta-feira:", dados[4])

# 2.2 Faturamento de todos os dias (coluna 1)
print("2.2 Faturamento de todos os dias:", dados[:, 1])

# 2.3 Quantidade de reclamações (coluna 4)
print("2.3 Quantidade de reclamações:", dados[:, 4])

# 2.4 Dados de Quinta-feira a Domingo (fatiamento de linhas de 3 em diante)
print("2.4 Dados de Quinta a Domingo:\n", dados[3:])

# 2.5 Apenas as colunas: Pedidos, Faturamento e Clientes (colunas 0, 1 e 2)
print("2.5 Colunas Pedidos, Faturamento e Clientes:\n", dados[:, :3])

# ==========================================
# Parte 3 — Operações numéricas
# ==========================================
print("\n--- Parte 3 — Operações numéricas ---")
total_pedidos = np.sum(dados[:, 0])
total_faturamento = np.sum(dados[:, 1])
total_clientes = np.sum(dados[:, 2])
media_pedidos = np.mean(dados[:, 0])
media_faturamento = np.mean(dados[:, 1])
media_reclamacoes = np.mean(dados[:, 4])

print(f"3.1 Total de pedidos durante a semana: {total_pedidos:.0f}")
print(f"3.2 Faturamento total: R$ {total_faturamento:,.2f}")
print(f"3.3 Total de clientes: {total_clientes:.0f}")
print(f"3.4 Média diária de pedidos: {media_pedidos:.2f}")
print(f"3.5 Média diária de faturamento: R$ {media_faturamento:,.2f}")
print(f"3.6 Média diária de reclamações: {media_reclamacoes:.2f}")

# ==========================================
# Parte 4 — Análise por dia
# ==========================================
print("\n--- Parte 4 — Análise por dia ---")
# 4.1 Pedidos
idx_max_pedidos = np.argmax(dados[:, 0])
idx_min_pedidos = np.argmin(dados[:, 0])
print(f"4.1 Maior número de pedidos: {dias[idx_max_pedidos]} ({dados[idx_max_pedidos, 0]:.0f} pedidos)")
print(f"    Menor número de pedidos: {dias[idx_min_pedidos]} ({dados[idx_min_pedidos, 0]:.0f} pedidos)")

# 4.2 Faturamento
idx_max_fat = np.argmax(dados[:, 1])
idx_min_fat = np.argmin(dados[:, 1])
print(f"4.2 Maior faturamento: {dias[idx_max_fat]} (R$ {dados[idx_max_fat, 1]:,.2f})")
print(f"    Menor faturamento: {dias[idx_min_fat]} (R$ {dados[idx_min_fat, 1]:,.2f})")

# 4.3 Amplitude do faturamento
amplitude_fat = np.max(dados[:, 1]) - np.min(dados[:, 1])
print(f"4.3 Amplitude do faturamento: R$ {amplitude_fat:,.2f}")

# ==========================================
# Parte 5 — Máscaras e filtros
# ==========================================
print("\n--- Parte 5 — Máscaras e filtros ---")
# 5.1 Faturamento > 12000
mask_fat_12k = dados[:, 1] > 12000
print("5.1 Dias com faturamento > R$ 12.000:", dias[mask_fat_12k])

# 5.2 Pedidos > 150
mask_ped_150 = dados[:, 0] > 150
print("5.2 Dias com mais de 150 pedidos:", dias[mask_ped_150])

# 5.3 Reclamações > 4
mask_rec_4 = dados[:, 4] > 4
print("5.3 Dias com mais de 4 reclamações:", dias[mask_rec_4])

# 5.4 Faturamentos dos dias com mais de 150 pedidos
print("5.4 Faturamentos dos dias com mais de 150 pedidos:", dados[mask_ped_150, 1])

# 5.5 Dias com faturamento abaixo da média semanal
dias_abaixo_media = np.sum(dados[:, 1] < media_faturamento)
print("5.5 Quantidade de dias com faturamento abaixo da média:", dias_abaixo_media)

# ==========================================
# Parte 6 — Classificação dos dias
# ==========================================
print("\n--- Parte 6 — Classificação dos dias ---")
classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

# 6.1 Apresentação da classificação
print("6.1 Classificação de cada dia:")
print(np.column_stack((dias, classificacao)))

# 6.2 Quantidade de dias de Alto desempenho
qtd_alto_desempenho = np.sum(classificacao == "Alto desempenho")
print(f"6.2 Quantidade de dias classificados como 'Alto desempenho': {qtd_alto_desempenho}")

# ==========================================
# Parte 7 — Estatística
# ==========================================
print("\n--- Parte 7 — Estatística do Faturamento ---")
fat = dados[:, 1]
print(f"Média do faturamento: R$ {np.mean(fat):,.2f}")
print(f"Mediana do faturamento: R$ {np.median(fat):,.2f}")
print(f"Desvio padrão do faturamento: R$ {np.std(fat):,.2f}")
print(f"Mínimo do faturamento: R$ {np.min(fat):,.2f}")
```
