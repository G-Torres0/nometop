```python
import pandas as pd
```

```python
#Importando um arquivo .csv dos dados de temperatura dos municípios
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
df
```

```python
cidade_busca = 'santa rosa'
cidade_df = df[df['nome_municipio'].str.lower() == cidade_busca.lower()]
```

```python
print("=== TODAS AS INFORMAÇÕES DA CIDADE ===")
# Transpõe (.T) a linha para exibir todas as 91 colunas verticalmente
print(cidade_df.T)
```

```python
cols_temp = [
    'temp_media_jan', 'temp_media_fev', 'temp_media_mar', 'temp_media_abr',
    'temp_media_mai', 'temp_media_jun', 'temp_media_jul', 'temp_media_ago',
    'temp_media_set', 'temp_media_out', 'temp_media_nov', 'temp_media_dez'
]

# Seleciona as colunas de temperatura do município
temps = cidade_df[cols_temp].iloc[0]

mes_temp_min = temps.idxmin()
valor_temp_min = temps.min()

print(f"\nMês de temperatura mínima: {mes_temp_min} com {valor_temp_min}°C")
```

```python


cols_def = [
    'def_jan', 'def_fev', 'def_mar', 'def_abr', 'def_mai', 'def_jun',
    'def_jul', 'def_ago', 'def_set', 'def_out', 'def_nov', 'def_dez'
]

deficits = cidade_df[cols_def].iloc[0]

mes_def_max = deficits.idxmax()
valor_def_max = deficits.max()

print(f"\nMês de maior déficit hídrico: {mes_def_max} com {valor_def_max} mm")
```

```python
estacoes = {
    'Verão': ['temp_media_dez', 'temp_media_jan', 'temp_media_fev'],
    'Outono': ['temp_media_mar', 'temp_media_abr', 'temp_media_mai'],
    'Inverno': ['temp_media_jun', 'temp_media_jul', 'temp_media_ago'],
    'Primavera': ['temp_media_set', 'temp_media_out', 'temp_media_nov']
}
print("\nMédia de temperatura por estação do ano:")
for estacao, meses in estacoes.items():
    media_estacao = temps[meses].mean()
    print(f"- {estacao}: {media_estacao:.2f}°C")
```
