```python
"""
Tema: Análise da Arrecadação de Impostos Federais por Estado e Região do Brasil.
Fonte: Dados Abertos da Receita Federal / Governo Federal.

Pergunta de Pesquisa:
Como a arrecadação federal se distribui entre os estados e regiões do Brasil,


Objetivos e Informações a Obter:
- Calcular o total arrecadado por Estado (UF) e por Região geográfica.
- Identificar os Top 5 estados de maior e menor arrecadação.

"""
```

```python



# OBTENÇÃO E EXPLORAÇÃO DOS DADOS
import pandas as pd

# Caminho do arquivo na pasta de dados do Colab
caminho_csv = "sample_data/arrecadacao-estado.csv"

# Leitura do arquivo CSV
try:
    df_arrecadacao = pd.read_csv(caminho_csv, sep=";", encoding="latin1")
except Exception:
    df_arrecadacao = pd.read_csv(caminho_csv, sep=",", encoding="latin1")

#  Visualizar as 5 primeiras linhas
print("--- Primeiras 5 linhas do dataset ---")
display(df_arrecadacao.head())

#  Verificar quantidade de linhas e colunas
linhas, colunas = df_arrecadacao.shape
print(f"\nDimensões do Dataset: {linhas} registros (linhas) e {colunas} colunas.")

#  Informações de tipos de dados e memória
print("\n--- Informações Estruturais do DataFrame ---")
df_arrecadacao.info()



# Visualizar as 5 primeiras linhas
print("--- Primeiras 5 linhas do dataset ---")
display(df_arrecadacao.head())

#  Verificar quantidade de linhas e colunas
linhas, colunas = df_arrecadacao.shape
print(f"\nDimensões do Dataset: {linhas} registros (linhas) e {colunas} colunas.")

# . Informações de tipos de dados e memória
print("\n--- Informações Estruturais do DataFrame ---")
df_arrecadacao.info()



#: PREPARAÇÃO E LIMPEZA DOS DADOS

# Verificar valores nulos por coluna
print("--- Valores Nulos Por Coluna ---")
print(df_arrecadacao.isnull().sum())

#  Dicionário de Mapeamento das UFs para as Regiões (P
mapeamento_regioes = {
    'SP': 'Sudeste', 'RJ': 'Sudeste', 'MG': 'Sudeste', 'ES': 'Sudeste',
    'PR': 'Sul', 'SC': 'Sul', 'RS': 'Sul',
    'DF': 'Centro-Oeste', 'GO': 'Centro-Oeste', 'MT': 'Centro-Oeste', 'MS': 'Centro-Oeste',
    'BA': 'Nordeste', 'PE': 'Nordeste', 'CE': 'Nordeste', 'MA': 'Nordeste',
    'PB': 'Nordeste', 'RN': 'Nordeste', 'AL': 'Nordeste', 'SE': 'Nordeste', 'PI': 'Nordeste',
    'AM': 'Norte', 'PA': 'Norte', 'RO': 'Norte', 'TO': 'Norte',
    'AC': 'Norte', 'AP': 'Norte', 'RR': 'Norte'
}

#  Normalizar nomes das colunas (remover espaços e ajustar caixa)
df_arrecadacao.columns = [col.strip().title() for col in df_arrecadacao.columns]
col_uf = 'Uf' if 'Uf' in df_arrecadacao.columns else df_arrecadacao.columns[0]
col_valor = 'Valor' if 'Valor' in df_arrecadacao.columns else df_arrecadacao.columns[-1]

#  Converter coluna de valor para float
if df_arrecadacao[col_valor].dtype == 'object':
    df_arrecadacao[col_valor] = (
        df_arrecadacao[col_valor]
        .astype(str)
        .str.replace('.', '', regex=False)
        .str.replace(',', '.', regex=False)
        .astype(float)
    )

#Remover linhas sem UF/Valor e criar coluna de Região
df_limpo = df_arrecadacao.dropna(subset=[col_uf, col_valor]).copy()
df_limpo['Regiao'] = df_limpo[col_uf].map(mapeamento_regioes)

print("\n--- Dataset Tratado e Pronto para Análise ---")
display(df_limpo.head())




# Total Geral Arrecadado
total_nacional = df_limpo[col_valor].sum()
print(f"Total Arrecadado no Brasil: R$ {total_nacional:,.2f}\n")

# Agrupamento por Região com pd.groupby()
arrecadacao_regiao = df_limpo.groupby('Regiao')[col_valor].sum().reset_index()
arrecadacao_regiao['Participacao_%'] = (arrecadacao_regiao[col_valor] / total_nacional) * 100
arrecadacao_regiao = arrecadacao_regiao.sort_values(by=col_valor, ascending=False)
print("--- Arrecadação por Região ---")
display(arrecadacao_regiao)

#Agrupamento por Estado (UF)
arrecadacao_uf = df_limpo.groupby([col_uf, 'Regiao'])[col_valor].sum().reset_index()
arrecadacao_uf['Participacao_%'] = (arrecadacao_uf[col_valor] / total_nacional) * 100
top_5_maiores = arrecadacao_uf.sort_values(by=col_valor, ascending=False).head(5)
top_5_menores = arrecadacao_uf.sort_values(by=col_valor, ascending=True).head(5)

print("\n--- TOP 5 Maiores Arrecadações por UF ---")
display(top_5_maiores)

print("\n--- TOP 5 Menores Arrecadações por UF ---")
display(top_5_menores)







```
