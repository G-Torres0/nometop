```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# 1 e 2. Buscar o dataset e carregar no Pandas
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=';')

# 3. Criar coluna de classificação
def classificar(nota):
    if nota <= 4:
        return 'Baixa'
    elif nota <= 6:
        return 'Média'
    return 'Alta'

df['classificacao'] = df['quality'].apply(classificar)

# 4 e 5. Separar dados de características (X) e alvo (y)
X = df.drop(columns=['quality', 'classificacao'])
y = df['classificacao']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 6. Treinar KNN (K inicial = 5)
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# 7. Conferir acurácia
y_pred = knn.predict(X_test)
print("Acurácia inicial (K=5):", accuracy_score(y_test, y_pred))

# 8. Testar outros valores de vizinhos (K)
print("\nTestando outros valores de K:")
for k in range(1, 15):
    modelo = KNeighborsClassifier(n_neighbors=k)
    modelo.fit(X_train, y_train)
    acc = accuracy_score(y_test, modelo.predict(X_test))
    print(f"K = {k} | Acurácia = {acc:.4f}")
```
