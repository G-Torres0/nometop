```python
import random

tentativas = 0
listadetentativas = []
num = random.randint(0, 15)

while True:
    digitado = input("QUAL O SEU CHUTE (ou digite sair): ")

    if digitado == "sair":
        break

    digitado = int(digitado)

    listadetentativas.append(digitado)
    tentativas += 1

    if digitado == num:
        print("Você acertou!")
        with open("historico.txt", "a") as f:
         f.write(f"Número: {num} - Tentativas: {tentativas}\n")
         num = random.randint(1, 15)
        tentativas = 0
else:
        print("Tente novamente!")


with open("historico.txt", "a") as f:
    f.write(f"Número: {num} - Tentativas: {tentativas}\n")


print("HISTÓRICO ")

with open("historico.txt", "r") as f:
    print(f.read())


```

```python
import csv

lista_compras = []
total = 0

while True:
    item = input("Digite o nome do produto (ou 'sair' para finalizar): ")

    if item.lower() == "sair":
        break

    quantidade = int(input("Digite a quantidade: "))
    valor = float(input("Digite o valor unitário: "))

    subtotal = quantidade * valor
    total += subtotal

    lista_compras.append([item, quantidade, valor, subtotal])

print("\n--- LISTA DE COMPRAS ---")

for compra in lista_compras:
    print(f"{compra[0]} - {compra[1]} x R$ {compra[2]:.2f} = R$ {compra[3]:.2f}")

print(f"\nValor total das compras: R$ {total:.2f}")

with open("lista_compras.csv", "w", newline="", encoding="utf-8") as arquivo:
    escritor = csv.writer(arquivo)

    escritor.writerow(["Produto", "Quantidade", "Valor Unitário", "Subtotal"])

    for compra in lista_compras:
        escritor.writerow(compra)

    escritor.writerow([])
    escritor.writerow(["TOTAL", "", "", total])

print("Lista salva no arquivo lista_compras.csv")
```
