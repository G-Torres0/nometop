```python
nome =   (input  ("qual o nome do aluno: "))
notaum = float (input("qual a primeira nota "))
notadois = float (input("qual a segunda nota "))
notatres = float (input("qual a terceira nota"))
media = (notaum+ notadois+ notatres)/3
print ("    RELATORIO   ")
print ("nome do aluno:", nome)
print ("nota um :", notaum)
print( "nota dois ", notadois)
print ("nota tres :", notatres)
print (f"a media do aluno e :{media:.2f}")
if media >= 7:
 print ("o aluno foi aprovado ")
else:
 print ("o aluno reprovou")
```
