```python
import json
import sys
import pandas as pd
import requests

# API Pública gratuita e estável (Dados de Usuários / Clientes)
URL_API = "https://jsonplaceholder.typicode.com/users"


def coletar_dados_api(url: str) -> list:
  """Realiza a requisição GET na API pública e valida o retorno."""
  try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    dados = response.json()

    # Garante que a resposta seja uma lista válida de registros
    if not isinstance(dados, list):
      print(f"[ERRO] A API retornou um formato inesperado: {dados}")
      sys.exit(1)

    return dados
  except requests.exceptions.RequestException as e:
    print(f"[ERRO] Falha ao conectar à API: {e}")
    sys.exit(1)


def estruturar_dados(dados_brutos: list) -> list:
  """Extrai e padroniza apenas os campos necessários."""
  dados_processados = []

  for item in dados_brutos:
    dados_processados.append({
        "id": item.get("id"),
        "nome": item.get("name", "N/A"),
        "username": item.get("username", "N/A"),
        "email": item.get("email", "N/A"),
        "cidade": item.get("address", {}).get("city", "N/A"),
        "empresa": item.get("company", {}).get("name", "N/A"),
        "website": item.get("website", "N/A"),
    })

  # Ordena alfabeticamente pelo nome do usuário
  return sorted(dados_processados, key=lambda x: x["nome"])


def salvar_arquivos(dados: list, prefixo_arquivo: str = "dados_usuarios"):
  """Salva os dados coletados nos formatos JSON e CSV."""
  caminho_json = f"{prefixo_arquivo}.json"
  caminho_csv = f"{prefixo_arquivo}.csv"

  # Salvar JSON
  with open(caminho_json, "w", encoding="utf-8") as f_json:
    json.dump(dados, f_json, ensure_ascii=False, indent=4)

  # Salvar CSV via Pandas
  df = pd.DataFrame(dados)
  df.to_csv(caminho_csv, index=False, encoding="utf-8-sig")

  print("[SUCESSO] Coleta e salvamento concluídos com sucesso!")
  print(f"- Arquivo JSON: {caminho_json}")
  print(f"- Arquivo CSV: {caminho_csv}")
  print(f"- Total de registros coletados: {len(dados)}")


def main():
  print("Iniciando coleta automática de dados...")
  dados_brutos = coletar_dados_api(URL_API)
  dados_estruturados = estruturar_dados(dados_brutos)
  salvar_arquivos(dados_estruturados)


if __name__ == "__main__":
  main()
```
