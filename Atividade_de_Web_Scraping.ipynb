```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Configurações para evitar bloqueios de layout e detecção headless
chrome_options = Options()
chrome_options.add_argument("--headless")
chrome_options.add_argument("--no-sandbox")
chrome_options.add_argument("--disable-dev-shm-usage")
chrome_options.add_argument("--window-size=1920,1080")  # Define resolução Desktop para evitar overlays
chrome_options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36")

driver = webdriver.Chrome(options=chrome_options)

cidade = input("Digite o nome da cidade: ").strip()

try:
    driver.get("https://www.climatempo.com.br")
    wait = WebDriverWait(driver, 15)

    # 1. Localiza o campo de busca
    campo_busca = wait.until(
        EC.element_to_be_clickable((By.CSS_SELECTOR, 'input[placeholder*="Buscar"], input[type="search"]'))
    )
    campo_busca.clear()
    campo_busca.send_keys(cidade)

    # 2. Aguarda a lista de sugestões carregar
    primeiro_resultado = wait.until(
        EC.presence_of_element_located((By.CSS_SELECTOR, "ul[class*='search'] li, .list-autocomplete li, a[class*='item']"))
    )

    # 3. Força o clique via JavaScript para ignorar anúncios/banners sobrepostos
    driver.execute_script("arguments[0].click();", primeiro_resultado)

    # 4. Captura as temperaturas da página de destino
    temp_min = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="min-temp-1" or contains(@id, "min-temp")]'))
    ).text

    temp_max = wait.until(
        EC.presence_of_element_located((By.XPATH, '//*[@id="max-temp-1" or contains(@id, "max-temp")]'))
    ).text

    print(f"\nPrevisão para '{cidade}':")
    print(f"Temperatura Mínima: {temp_min}")
    print(f"Temperatura Máxima: {temp_max}")

except Exception as erro:
    print(f"\nOcorreu um erro ao buscar a previsão: {erro}")

finally:
    driver.quit()
```
